Imagine you are working as a Data Scientist supporting a drug development team at Pfizer. The team is running stability tests on several drugs, recording temperature and pressure at hourly intervals every day. These readings determine whether a drug stays stable under different conditions, so getting the data right is critical.

The drug-testing team needs to prepare raw hourly readings for analysis - calculate daily temperature and pressure summaries, classify temperatures into bands, identify missing readings, and examine selected drugs. How can we restructure and process the data to support these tasks?

#Handle Missing Values

```python
#import pandas as pd

data = pd.read_csv('Pfizer_1.csv')

data_melt = pd.melt(data, id_vars=['Date', 'Drug_Name', 'Parameter'],
                     var_name="time", value_name='reading')

data_tidy = data_melt.pivot(index=['Date', 'time', 'Drug_Name'],
                            columns='Parameter', values='reading')
data_tidy = data_tidy.reset_index()
data_tidy.columns.name = None
```

#NaN vs None

```python
type(None)
```

```python
import numpy as np

type(np.nan)
```

```python
pd.Series([1,np.nan,2,None])
```

For a numeric Series, Pandas changes None to NaN. Both become the same thing internally - every value in the output prints as NaN, and you can no longer tell which ones started out as None.

```python
pd.Series(["1","np.nan","2",np.nan])
```

Three of these four values are strings - "1", "np.nan" (yes, literally the text), and "2" - and only the unquoted np.nan is an actual missing value, so it is the one entry that prints as NaN.

```python
pd.Series(["1","2",None,np.nan])
```
