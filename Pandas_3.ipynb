#Group based Filtering

Find the directors who have atleast one movie with budget greater than 100 M

```python
data_dir_budget = data.groupby('director_name')['budget'].max().reset_index()
data_dir_budget.head()
```

```python
names = data_dir_budget.loc[data_dir_budget['budget'] > 100000000, "director_name"]
names
```

```python
data.loc[data['director_name'].isin(names)]
```

```python
#Alternate approach
```

```python
data.groupby('director_name').filter(lambda x: x['budget'].max() > 100000000)
```

#Filter Groups on Multiple conditions

Filter out directors who have an avg budget > 50 M and total revenue > 100M

```python
data.groupby('director_name').filter(lambda x: x['budget'].mean() > 50000000 and x['revenue'].sum() >= 100000000)
```

#Group based Apply()

```python
def func(x):
  x["risky"] = x["budget"] - x["revenue"].mean() >=0
  return x

data_risky = data.groupby('director_name', group_keys=False).apply(func)
data_risky
```

```python
data.groupby('director_name', group_keys=True).apply(func) #default = True #ignore group keys
```
